```python
import pandas as pd

df = pd.read_parquet("./data/Cell_Painting_all_compounds_all_doses.parquet")
cmpds = pd.read_csv("data/compounds_with_assay_data.csv")

df
```

```python
cmpds["Metadata_pert_id"] = cmpds["BROAD_ID"].str[:13]
```

```python
df = df[df.Metadata_pert_id.isin(cmpds.Metadata_pert_id.unique())].reset_index(drop=True)
df
```

```python
df.Metadata_pert_id.unique()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(60, 6))
sns.boxplot(
    data=df,
    x='Metadata_Plate',
    y='Cells_Number_Object_Number',
    color='skyblue',
    showfliers=False  # optional: hide outliers for cleaner plot
)
plt.xticks(rotation=90)
plt.title('Variation of Cells_Number_Object_Number Across Plates')
plt.xlabel('Plate ID')
plt.ylabel('Cells_AreaShape_Extent')

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# List of features to compare (change or add your own)
features = [
    'Cells_Number_Object_Number',
    'Cells_RadialDistribution_RadialCV_Mito_1of4',
    'Cells_Texture_Contrast_AGP_10_0',
    'Cytoplasm_Granularity_10_ER',
    'Cytoplasm_RadialDistribution_RadialCV_RNA_4of4'
    
]

# Compute mean per plate for each feature and reshape to long format
means_list = []
for feat in features:
    means = df.groupby('Metadata_Plate')[feat].std().reset_index()
    means = means.rename(columns={feat: 'mean_value'})
    means['feature'] = feat
    means_list.append(means)

means_long = pd.concat(means_list)

# Plot boxplot comparing features
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=means_long,
    x='feature',
    y='mean_value',
    palette='pastel'
)

plt.title('Variation in Mean Feature Values per Plate')
plt.xlabel('Feature')
plt.ylabel('Mean Value per Plate')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
