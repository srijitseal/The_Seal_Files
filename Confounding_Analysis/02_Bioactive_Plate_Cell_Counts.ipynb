```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Combine datasets
bioactive_df['Group'] = 'Bioactive'
inactive_df['Group'] = 'Inactive'
feature = 'Cells_Number_Object_Number'

combined_df = pd.concat([
    bioactive_df[['Metadata_Plate', feature, 'Group']],
    inactive_df[['Metadata_Plate', feature, 'Group']]
])

# Step 2: Compute per-plate mean values
mean_by_plate_group = combined_df.groupby(['Metadata_Plate', 'Group'])[feature].mean().unstack()

# Step 3: Compute the difference
mean_by_plate_group['Difference'] = mean_by_plate_group['Bioactive'] - mean_by_plate_group['Inactive']

# Step 4: Sort plates by difference
mean_by_plate_group_sorted = mean_by_plate_group.sort_values('Difference')

# Step 5: Plot
plt.figure(figsize=(20, 6), dpi=200)
sns.barplot(
    x=mean_by_plate_group_sorted.index,
    y=mean_by_plate_group_sorted['Difference'],
    palette='coolwarm'
)
plt.axhline(0, color='gray', linestyle='--')
plt.xticks(rotation=90)
plt.title(f'Mean Difference in {feature} Between Bioactive and Inactive (Per Plate)')
plt.xlabel('Plate ID')
plt.ylabel('Mean(Bioactive) - Mean(Inactive)')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Combine and compute as before
bioactive_df['Group'] = 'Bioactive'
inactive_df['Group'] = 'Inactive'
feature = 'Cells_Number_Object_Number'

combined_df = pd.concat([
    bioactive_df[['Metadata_Plate', feature, 'Group']],
    inactive_df[['Metadata_Plate', feature, 'Group']]
])

# Compute per-plate group mean
mean_by_plate_group = combined_df.groupby(['Metadata_Plate', 'Group'])[feature].mean().unstack()

# Compute the difference
mean_by_plate_group['Difference'] = mean_by_plate_group['Bioactive'] - mean_by_plate_group['Inactive']

# Sort differences
sorted_diffs = mean_by_plate_group['Difference'].sort_values()

# Plot
plt.figure(figsize=(10, 5), dpi=300)
sns.lineplot(data=sorted_diffs.values, marker='o', linestyle='-')

# Aesthetics: remove top and right spines
sns.despine(top=True, right=True)

# Add zero baseline
plt.axhline(0, color='gray', linestyle='--')

# Labels
plt.title(f'Sorted Per-Plate Difference in {feature} (Bioactive - Inactive)')
plt.xlabel('Plates (sorted)')
plt.ylabel('Mean Difference')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statannot import add_stat_annotation

# Define the feature
feature = 'Cells_Number_Object_Number'

# Label groups
bioactive_df['Group'] = 'Bioactive'
inactive_df['Group'] = 'Inactive'

# Combine and subset relevant columns
combined_df = pd.concat([
    bioactive_df[['Metadata_Plate', feature, 'Group']],
    inactive_df[['Metadata_Plate', feature, 'Group']]
])

# Compute per-plate mean
plate_means = (
    combined_df
    .groupby(['Metadata_Plate', 'Group'])[feature]
    .mean()
    .reset_index()
)

# Convert group labels to string (if needed)
plate_means['Group'] = plate_means['Group'].astype(str)
order = ['Inactive', 'Bioactive']
palette = ["#1f77b4", "#ff7f0e"]

# Plot
plt.figure(figsize=(6, 6), dpi=300)

sns.violinplot(
    data=plate_means,
    x='Group',
    y=feature,
    palette=palette,
    inner=None,
    cut=0,
    order=order
)

sns.swarmplot(
    data=plate_means,
    x='Group',
    y=feature,
    color='k',
    alpha=0.8,
    size=2,
    order=order
)

# Add stat annotation
add_stat_annotation(
    ax=plt.gca(),
    data=plate_means,
    x='Group',
    y=feature,
    order=order,
    box_pairs=[("Inactive", "Bioactive")],
    test="Mann-Whitney",
    text_format="star",
    loc="outside",
    verbose=1,
)

# Final formatting
sns.despine(top=True, right=True)
plt.xlabel('')
plt.ylabel(f'Mean {feature}')
plt.tight_layout()
plt.show()
```

```python
# Pivot the table so each plate has two columns: one for each group
pivot_df = plate_means.pivot(index='Metadata_Plate', columns='Group', values=feature)

# Drop rows (plates) with missing values in either group
paired_plates = pivot_df.dropna()

# Count how many plates have both groups
num_paired_plates = len(paired_plates)

# Count how many of those have bioactive mean < inactive mean
num_bioactive_lower = (paired_plates['Bioactive'] < paired_plates['Inactive']).sum()

# Print results
print(f"Number of plates with both groups: {num_paired_plates}")
print(f"Number of plates where bioactive mean < inactive mean: {num_bioactive_lower}")
```
