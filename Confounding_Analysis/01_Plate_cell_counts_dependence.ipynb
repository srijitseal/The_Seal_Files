```python
import pandas as pd
import numpy as np

# Assume df contains your Cell Painting data with columns:
# - 'Metadata_Plate' for plate IDs
# - feature columns (~1800 of them)

# List of feature columns (exclude metadata columns)
feature_cols = [c for c in df.columns if c not in ['Metadata_Plate', 'Metadata_Well', 'Metadata_Assay_Plate_Barcode', 'Metadata_mmoles_per_liter']]

# Compute per-plate std for all features
per_plate_std = df.groupby('Metadata_Plate')[feature_cols].std()

# Average (mean) std deviation of each feature across plates
feature_variation_summary = per_plate_std.mean(axis=0).reset_index()
feature_variation_summary.columns = ['Feature', 'Mean_Plate_StdDev']

# Sort by variation descending
feature_variation_summary = feature_variation_summary.sort_values('Mean_Plate_StdDev', ascending=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(feature_variation_summary['Mean_Plate_StdDev'], bins=50, color='skyblue')
plt.xlabel('Mean Per-Plate Std Deviation')
plt.ylabel('Number of Features')
plt.title('Distribution of Per-Plate Variation Across All Features')
plt.xscale('log')  # <-- Add this for log scale on x-axis
plt.tight_layout()
plt.show()
```

```python
# Features of interest
feat1 = 'Cells_Number_Object_Number'
feat2 = 'Cells_RadialDistribution_RadialCV_Mito_1of4'

plt.figure(figsize=(10, 6))
sns.histplot(feature_variation_summary['Mean_Plate_StdDev'], bins=50, kde=True, color='lightgray')

# Highlight feat1 and feat2
for feat, color in zip([feat1, feat2], ['red', 'blue']):
    val = feature_variation_summary.loc[feature_variation_summary['Feature'] == feat, 'Mean_Plate_StdDev'].values[0]
    plt.axvline(val, color=color, linestyle='--', label=f'{feat} ({val:.2f})')

plt.xlabel('Mean Per-Plate Std Deviation')
plt.ylabel('Number of Features')
plt.title('Per-Plate Variation Distribution with Highlighted Features')
plt.xscale('log')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
top20 = feature_variation_summary.head(50)

plt.figure(figsize=(12, 6), dpi=300)
sns.barplot(data=top20, x='Feature', y='Mean_Plate_StdDev')
sns.despine(top=True, right=True)

plt.xticks(rotation=90)
plt.ylabel('Mean Per-Plate Std Deviation')
plt.xlabel('Feature')
plt.yscale('log')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# List of features to compare (change or add your own)
features = [
    'Cells_Number_Object_Number',
    'Cells_RadialDistribution_RadialCV_Mito_1of4',
    'Cells_Texture_Contrast_AGP_10_0',
    'Cytoplasm_Granularity_10_ER',
    'Cytoplasm_RadialDistribution_RadialCV_RNA_4of4'
    
]

# Compute mean per plate for each feature and reshape to long format
means_list = []
for feat in features:
    means = df.groupby('Metadata_Plate')[feat].std().reset_index()
    means = means.rename(columns={feat: 'mean_value'})
    means['feature'] = feat
    means_list.append(means)

means_long = pd.concat(means_list)

# Plot boxplot comparing features
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=means_long,
    x='feature',
    y='mean_value',
    palette='pastel'
)

plt.title('Variation in Mean Feature Values per Plate')
plt.xlabel('Feature')
plt.ylabel('Mean Value per Plate')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
