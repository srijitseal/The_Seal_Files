```python
import pandas as pd

df = pd.read_csv("../Cell_Painting_Median_Features_Median_doses.csv")
df
```

```python
import pandas as pd
from rdkit import Chem

# Function to convert InChI to SMILES
def inchi_to_smiles(inchi):
    try:
        mol = Chem.inchi.MolFromInchi(inchi)
        if mol:
            return Chem.MolToSmiles(mol)
    except:
        return None

# Apply conversion to the column
df['SMILES'] = df['InChICode_standardised'].apply(inchi_to_smiles)
df
```

```python
df = df[["InChICode_standardised", "SMILES", "Cells_Number_Object_Number"]]
df
```

```python
# Ensure numeric format
df['Cells_Number_Object_Number'] = pd.to_numeric(df['Cells_Number_Object_Number'], errors='coerce')

# Drop missing values
df = df.dropna(subset=['Cells_Number_Object_Number'])

# Calculate the 5th percentile (left tail)
left_tail_cutoff = df['Cells_Number_Object_Number'].quantile(0.05)

# Get compounds in the bottom 5% (left tail)
left_tail_df = df[df['Cells_Number_Object_Number'] <= left_tail_cutoff]

print("Compounds in the left tail (lowest 5%):")
print(left_tail_df[['InChICode_standardised', 'SMILES', 'Cells_Number_Object_Number']])
```

```python
left_tail_df = left_tail_df.sort_values("Cells_Number_Object_Number").reset_index(drop=True)
left_tail_df
```

```python
left_tail_df.to_csv("Bray_low_cell_count.csv", index=False)
```
