```python
import pandas as pd
```

```python
df = pd.read_csv("Cell_Painting_Median_Features_Median_doses.csv")
```

```python
df = df.drop("index", axis=1)
df
```

```python
# Ensure numeric format
df['Cells_Number_Object_Number'] = pd.to_numeric(df['Cells_Number_Object_Number'], errors='coerce')

# Drop missing values
df = df.dropna(subset=['Cells_Number_Object_Number'])

# Calculate the 5th percentile (left tail)
left_tail_cutoff = df['Cells_Number_Object_Number'].quantile(0.05)

# Get compounds in the bottom 5% (left tail)
left_tail_df = df[df['Cells_Number_Object_Number'] <= left_tail_cutoff]

print("Compounds in the left tail (lowest 5%):")
print(len(left_tail_df))
```

```python
df = df[~df.InChICode_standardised.isin(left_tail_df.InChICode_standardised.to_list())]
df
```

```python
import pandas as pd
from pandarallel import pandarallel

# Initialize pandarallel
pandarallel.initialize()

# Assuming you have a large dataframe 'df'
# Define the target column
target_column = 'Cells_Number_Object_Number'

# Function to calculate correlation
def calculate_correlation(column):
    if column == target_column:
        return 1
    subset = df[[target_column, column]].dropna()
    if subset.shape[0] > 1:
        return subset.corr().iloc[0, 1]
    else:
        return None

# Apply the function in parallel
correlations = df.columns[1:].to_series().parallel_apply(calculate_correlation)

# Filter columns with high correlation (absolute value > 0.8)
highly_correlated_columns = correlations[correlations > 0.8].dropna().index.tolist()

# Print the highly correlated columns
print(highly_correlated_columns)
```

```python
df_cleaned = df.drop(columns=highly_correlated_columns, axis=1) 
```

```python
df_cleaned = df_cleaned.reset_index(drop=True)
df_cleaned 
```

```python
df_cleaned.to_parquet("Cell_Painting_no_cellcount_confounder_dataset/Bray_etal_Cell_Painting_no_cellcount_confounder_dataset.parquet", index=False)
```
